```python
words = open('names.txt', 'r').read().splitlines()
```

```python
words[:10]
```

```python
len(words)
```

```python
# 接着我们看一下Bigram如何实现两个字符的逐步输出，
# 因为是每个字符去预测下一个字符
# zip 的工作就是把两个序列按位置一一配对
# w: [ 'e',   'm',   'm',   'a' ]
# w[1:]:   [ 'm',   'm',   'a' ]
b = {}
for w in words:
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1, ch2 in zip(chs, chs[1:]):
        # 看字母的出现的规律，用字典来计数
        bigram = (ch1,ch2)
        b[bigram] = b.get(bigram, 0) + 1
        # print(ch1, ch2)
```

```python
# 接下来对b进行计数排序
sorted(b.items(), key = lambda kv : -kv[1])
```
